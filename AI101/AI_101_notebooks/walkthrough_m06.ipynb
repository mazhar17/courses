## Step 1. Inference with fixed parameters

Lesson 6.1 says the intercept is 2 and the slope is 3. At x = 4, inference gives ŷ = 2 + 3 × 4 = 14. Predicting uses the parameters. It does not change them.

````python
# Step 1: inference with a model whose parameters are already known
b0 = 2                    # intercept
b1 = 3                    # slope

x_new = 4
y_hat = b0 + b1 * x_new
print("Prediction at x = 4:", y_hat)

x_new = 10                # a different input; the parameters stay as they are
print("Prediction at x = 10:", b0 + b1 * x_new)
print("Parameters after predicting:", b0, b1)
````

**Expected output**

```
Prediction at x = 4: 14
Prediction at x = 10: 32
Parameters after predicting: 2 3
```

The prediction at x = 4 is 14, as in the lesson. A new input gives a new prediction, but b0 and b1 are unchanged.

## Step 2. Fit with scikit-learn, then predict

Now the computer finds the parameters. X must be a table with one row per example and one column per feature (Module 3), so each x value sits in its own list. The `fit` call learns the intercept and slope from the examples. The `predict` call only applies them.

````python
# Step 2: training (fit) and inference (predict)
import numpy as np
from sklearn.linear_model import LinearRegression

X_train = np.array([[1], [2], [3], [4], [5]])    # shape (5, 1): 5 examples, 1 feature
y_train = np.array([5, 8, 11, 14, 17])

model = LinearRegression()
model.fit(X_train, y_train)                      # training: learn the parameters
print(f"Fitted intercept: {model.intercept_:.2f}")
print(f"Fitted slope: {model.coef_[0]:.2f}")

print(f"Prediction at x = 4: {model.predict(np.array([[4]]))[0]:.2f}")
print(f"Intercept after predicting: {model.intercept_:.2f}")
````

**Expected output**

```
Fitted intercept: 2.00
Fitted slope: 3.00
Prediction at x = 4: 14.00
Intercept after predicting: 2.00
```

Fitting recovers the intercept 2 and the slope 3 that generated the points. The prediction at x = 4 is 14, and predicting did not change the intercept. `model.predict` returns an array, so `[0]` picks its first value.

## Step 3. Fitting again can change the parameters

Lesson 6.1 says fitting again may change the intercept or slope. Here you fit a new model on different examples that follow y = 1 + 4x. Retraining is a separate operation with its own data.

````python
# Step 3: fit a new model on different examples
y_other = np.array([5, 9, 13, 17, 21])           # y = 1 + 4x

model_2 = LinearRegression()
model_2.fit(X_train, y_other)
print(f"New intercept: {model_2.intercept_:.2f}")
print(f"New slope: {model_2.coef_[0]:.2f}")
print(f"First model still has slope: {model.coef_[0]:.2f}")
````

**Expected output**

```
New intercept: 1.00
New slope: 4.00
First model still has slope: 3.00
```

Different training examples gave different parameters. The first model keeps its own parameters, because you created a second model object. In a real project, retraining needs its own data and validation (Lesson 6.1).

## Step 4. The loss changes what fitting emphasizes

Lesson 6.2 says mean absolute error (MAE) averages the absolute residuals. Mean squared error (MSE) squares them before averaging, so large residuals count more. Here are four residuals. One is much larger than the others. They are invented, in an arbitrary unit.

````python
# Step 4: the same residuals, two different losses
residuals = [1, 1, 1, 10]

abs_total = 0
sq_total = 0
for r in residuals:
    abs_total += abs(r)
    sq_total += r ** 2

print("MAE:", abs_total / len(residuals))
print("MSE:", sq_total / len(residuals))
````

**Expected output**

```
MAE: 3.25
MSE: 25.75
```

One large residual of 10 dominates the MSE. It contributes 100 of the total 103 in the sum of squares. It contributes 10 of the total 13 in the sum of absolute values. The choice of loss changes what fitting emphasizes. MSE is not in the units of the output, but its square root is (Module 7).

## Step 5. Compare a model with a baseline

In Lesson 6.3 the supplied model has an MAE of 1.2 kW and the physical baseline has an MAE of 0.8 kW on the same held-out tests. Convert both to watts, then compare them.

````python
# Step 5: model versus baseline on the same held-out tests
ml_mae_kw = 1.2           # MAE of the ML model in kW
baseline_mae_kw = 0.8     # MAE of the physical baseline in kW

ml_mae_w = ml_mae_kw * 1000
baseline_mae_w = baseline_mae_kw * 1000
print(f"ML model MAE: {ml_mae_w:.0f} W")
print(f"Baseline MAE: {baseline_mae_w:.0f} W")

if ml_mae_w < baseline_mae_w:
    print("The ML model has the lower MAE on these tests.")
else:
    print("The baseline has the lower MAE, so the ML model has not shown an improvement.")
````

**Expected output**

```
ML model MAE: 1200 W
Baseline MAE: 800 W
The baseline has the lower MAE, so the ML model has not shown an improvement.
```

1.2 kW is 1,200 W and 0.8 kW is 800 W, as in Lesson 6.3. The baseline has the lower error. Lesson 6.3 says to report that result and not to force a case for ML. Cost, interpretability and evidence for the operating range also matter.
