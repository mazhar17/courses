# Read a pump table as X and y

**Objective.** Turn a small pump table into a feature matrix X and a target array y. Check their shapes, check that each row stays aligned with its target, and check that every feature exists at decision time.

**Supports.** Lesson 3.1 (sections “Read each row as an example”, “Separate features from targets” and “Preserve row alignment”) and Lesson 3.3 (section “Check when information becomes available”).

Run the cells from top to bottom. All data are synthetic and invented for teaching.


## About the data

Lesson 3.1 uses a pump table in which one row is one steady operating test. The columns are flow rate, pressure rise, shaft speed and measured electrical power. Here you build a table with four tests.

The four tests are synthetic. They were invented for teaching and are not measurements from a real pump.

```
Test  Q (m³/s)  Δp (Pa)  Speed (rad/s)  Measured power (W)
T1      0.020   100000            150                2800
T2      0.015    80000            130                1650
T3      0.025   120000            170                3950
T4      0.018    90000            140                2100
```


## Step 1. Store the table, with units in the names

Each test is one group of five values: the test identifier, flow rate, pressure rise, shaft angular speed and measured power. The units are written in the comments, so the numbers keep their meaning (Lesson 3.2).

````python
# Step 1: the table, one tuple per test (synthetic data)
# (test id, Q in m³/s, pressure rise in Pa, speed in rad/s, power in W)
tests = [
    ("T1", 0.020, 100000, 150, 2800),
    ("T2", 0.015,  80000, 130, 1650),
    ("T3", 0.025, 120000, 170, 3950),
    ("T4", 0.018,  90000, 140, 2100),
]
print("Number of tests:", len(tests))
````

**Expected output**

```
Number of tests: 4
```

`len(tests)` counts the items in the list. There are four samples, one per row.

## Step 2. The shapes in Lesson 3.1

Lesson 3.1 says X has shape (n, d) and y has shape (n,). Its example has 120 tests and three features. You can check the lesson’s numbers with plain arithmetic.

````python
# Step 2: the lesson's numbers, 120 tests and 3 features
n = 120                   # number of samples (rows)
d = 3                     # number of features (columns)
print("X has shape:", (n, d))
print("y has shape:", (n,))
print("Feature values in X:", n * d)
````

**Expected output**

```
X has shape: (120, 3)
y has shape: (120,)
Feature values in X: 360
```

This matches Lesson 3.1: 120 × 3 = 360 feature values and 120 target values. The comma in `(n,)` shows that y has one dimension, with n values.

## Step 3. Which columns may be features?

Lesson 3.3 says a valid predictor must exist when the decision is made. Suppose you want to predict the power for a planned operating point. Flow rate, pressure rise and shaft speed are set or known beforehand. The measured power is the target. A repair code that is only written after a failure is not available beforehand. The list below is invented for teaching.

````python
# Step 3: is each column available at decision time?
columns = [
    ("flow rate", True),
    ("pressure rise", True),
    ("shaft speed", True),
    ("repair code written after a failure", False),
]

for name, available in columns:
    if available:
        print("can be a feature:   ", name)
    else:
        print("NOT a valid feature:", name)
````

**Expected output**

```
can be a feature:    flow rate
can be a feature:    pressure rise
can be a feature:    shaft speed
NOT a valid feature: repair code written after a failure
```

A column can be physically meaningful and still invalid for prediction because it appears too late. This check is a thinking aid. Only you can decide, from the engineering question, what is available.

## Step 4. Build X and y and keep the rows aligned

Now you split each test into features and a target. The `for` loop handles one test at a time, so the features and the target in the same position always come from the same test. Then you turn the two lists into NumPy arrays.

````python
# Step 4: build X (features) and y (target), then check shapes and alignment
import numpy as np

X_rows = []               # one list of features per test
y_values = []             # one target per test
for test_id, q, dp, speed, power in tests:
    X_rows.append([q, dp, speed])
    y_values.append(power)

X = np.array(X_rows)
y = np.array(y_values)
print("X shape:", X.shape)
print("y shape:", y.shape)
print("Values in X:", X.size)
print("Row 2 of X:", X[2], "-> target", y[2])
````

**Expected output**

```
X shape: (4, 3)
y shape: (4,)
Values in X: 12
Row 2 of X: [2.5e-02 1.2e+05 1.7e+02] -> target 3950
```

X has 4 rows (tests) and 3 columns (features), so it holds 12 values. Row 2 counts from 0, so it is test T3. Its target is 3950 W, as in the table. NumPy prints the numbers in scientific notation, so 2.5e-02 means 0.025 m³/s. The units are not stored in the array, so you must keep them in your notes (Lesson 3.1).
