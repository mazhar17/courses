# Check an AI-assisted photovoltaic estimate

**Optional code walkthrough for AI 101 (Module 14).** Not graded. Quiz scores do not depend on it.

**Objective.** Calculate the photovoltaic (PV) power of Lesson 14.1 yourself. Use the result to catch a factor-of-1,000 unit error in a claimed answer. Then see why one power value cannot give daily energy.

**Supports.** Lesson 14.1 (sections “Verify a photovoltaic power estimate” and “Keep power and energy distinct”).

Run the cells from top to bottom. All data are synthetic and invented for teaching.


## About the data

The illustrative array of Lesson 14.1 has area A = 20 m², irradiance G = 800 W/m², module efficiency 0.20 and inverter efficiency 0.96. The simplified model is P_DC = η A G and P_AC = η_inv P_DC.

Suppose an assistant answers “3,200 kW”. You check that claim. The numbers are an authored teaching case. They are not from real equipment.

```
Area A:                    20 m²
Irradiance G:              800 W/m²
Module efficiency:         0.20  (dimensionless)
Inverter efficiency:       0.96  (dimensionless)
Claim to check:            3200 kW
```


## Step 1. Calculate the DC power yourself

Do your own calculation before you trust an answer. The unit check comes first: m² × W/m² gives W. The efficiencies have no unit.

````python
# Step 1: P_DC = efficiency * area * irradiance
efficiency = 0.20         # dimensionless
area = 20                 # m²
irradiance = 800          # W/m²
p_dc = efficiency * area * irradiance   # W
print(f"DC power: {p_dc:.1f} W")
print(f"DC power: {p_dc / 1000:.1f} kW")
````

**Expected output**

```
DC power: 3200.0 W
DC power: 3.2 kW
```

This matches Lesson 14.1: 3,200 W, which is 3.2 kW. Dividing by 1,000 converts watts to kilowatts.

## Step 2. Calculate the AC power

The inverter converts DC to AC. With inverter efficiency 0.96 the simplified AC estimate is P_AC = η_inv P_DC.

````python
# Step 2: P_AC = inverter efficiency * P_DC
inverter_efficiency = 0.96   # dimensionless
p_ac = inverter_efficiency * p_dc
print(f"AC power: {p_ac:.1f} W")
print(f"AC power: {p_ac / 1000:.3f} kW")
````

**Expected output**

```
AC power: 3072.0 W
AC power: 3.072 kW
```

This matches Lesson 14.1: 3,072 W, which is 3.072 kW. The estimate is simplified. It ignores temperature effects and other losses.

## Step 3. Test a claimed answer

Suppose an assistant writes “P_DC = 3,200 kW”. Convert the claim to watts and compare it with your own result. A ratio near 1,000 points to a unit error.

````python
# Step 3: compare a claimed value with your own calculation
claimed_kw = 3200
claimed_w = claimed_kw * 1000     # 1 kW = 1000 W
ratio = claimed_w / p_dc
print(f"Claimed DC power: {claimed_w} W")
print(f"Ratio claimed / calculated: {ratio:.1f}")
if ratio > 100 or ratio < 0.01:
    print("The claim is off by a large factor. Check the units first.")
else:
    print("The claim is the same order of magnitude as the calculation.")
````

**Expected output**

```
Claimed DC power: 3200000 W
Ratio claimed / calculated: 1000.0
The claim is off by a large factor. Check the units first.
```

The claim has the right digits but the wrong unit. It is 1,000 times too large, as Lesson 14.1 explains. A fluent answer can contain a correct-looking number with the wrong unit. This check only finds the large unit error. A claim that is close but wrong would pass it.

## Step 4. Power is not energy

Energy is the time integral of power. Lesson 14.1 allows one explicitly illustrative calculation: constant power for two hours only. It must not be presented as a daily yield. Energy in kWh is power in kW times time in hours.

````python
# Step 4: energy under an explicit constant-power assumption
p_ac_kw = p_ac / 1000     # kW
hours = 2                 # stated duration, constant power assumed
energy_kwh = p_ac_kw * hours
print(f"Energy in {hours} h at constant power: {energy_kwh:.3f} kWh")
print("This is not a daily yield. A daily value needs a time profile.")
````

**Expected output**

```
Energy in 2 h at constant power: 6.144 kWh
This is not a daily yield. A daily value needs a time profile.
```

3.072 kW × 2 h = 6.144 kWh, as in Lesson 14.1. The result is valid only for the two stated hours and the constant-power assumption. Sunshine changes during the day. One power value cannot give daily energy.

## Try it

Change a value and observe. In Step 4, change `hours = 2` to `hours = 3`. Then run the cell again.

**Expected result**

```
Energy in 3 h at constant power: 9.216 kWh
This is not a daily yield. A daily value needs a time profile.
```

Energy grows in proportion to the duration: 3.072 kW × 3 h = 9.216 kWh. The assumption that power stays constant for three hours is now less realistic. A longer duration does not make the estimate more correct.
