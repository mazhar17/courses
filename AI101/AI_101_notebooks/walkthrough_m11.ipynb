## About the data

All numbers come from the lessons. The neuron uses x₁ = 2, x₂ = 1, w₁ = 0.5, w₂ = −1 and b = 0.2. The update uses θ_old = 2, gradient 0.5 and learning rate η = 0.1. The epoch example has 1,000 training examples and batch size 50.

Lessons 11.1 and 11.2 treat these values as dimensionless. They are teaching numbers and not measurements from real equipment.

```
Forward example:  x1 = 2, x2 = 1, w1 = 0.5, w2 = -1, b = 0.2
Update example:   theta_old = 2, gradient = 0.5, learning rate = 0.1
```

## Step 1. The weighted sum of one neuron

Lesson 11.1 calculates the weighted sum z = w₁x₁ + w₂x₂ + b. First calculate the two products. Then add them and the bias. Applying the activation is a separate, later step.

````python
# Step 1: weighted sum of one neuron (dimensionless numbers)
x1 = 2
x2 = 1
w1 = 0.5
w2 = -1
b = 0.2

contribution1 = w1 * x1
contribution2 = w2 * x2
z = contribution1 + contribution2 + b
print(f"First contribution: {contribution1:.1f}")
print(f"Second contribution: {contribution2:.1f}")
print(f"Weighted sum z: {z:.1f}")
````

**Expected output**

```
First contribution: 1.0
Second contribution: -1.0
Weighted sum z: 0.2
```

This matches Lesson 11.1. The two contributions cancel, so z equals the bias, 0.2.

## Step 2. Apply the ReLU activation

ReLU is a = max(0, z). Negative values give zero. Positive values pass through unchanged. First you check the three examples from the lesson. Then you apply ReLU to the z from Step 1.

````python
# Step 2: ReLU, a = max(0, z)
def relu(z):
    return max(0, z)

for test_z in [-3, 0.2, 2]:
    print(f"ReLU({test_z}) = {relu(test_z)}")

a = relu(z)
print(f"Neuron output a: {a:.1f}")
````

**Expected output**

```
ReLU(-3) = 0
ReLU(0.2) = 0.2
ReLU(2) = 2
Neuron output a: 0.2
```

The three values match the lesson: −3 gives 0, 0.2 gives 0.2 and 2 gives 2. Because z = 0.2 is positive, ReLU leaves it unchanged. This is a forward calculation only. It does not calculate a gradient or change a weight.

## Step 3. One gradient-descent update

Lesson 11.2 gives the rule θ_new = θ_old − η∇L. Here the gradient is supplied by the lesson. The code does not derive it from a loss function.

````python
# Step 3: one update, theta_new = theta_old - learning_rate * gradient
theta_old = 2
gradient = 0.5            # supplied by the lesson
learning_rate = 0.1       # a training setting, not a fitted weight

adjustment = learning_rate * gradient
theta_new = theta_old - adjustment
print(f"Adjustment: {adjustment:.2f}")
print(f"New parameter: {theta_new:.2f}")
````

**Expected output**

```
Adjustment: 0.05
New parameter: 1.95
```

This matches Lesson 11.2: 0.1 × 0.5 = 0.05 and 2 − 0.05 = 1.95. The learning rate scales the gradient before it is subtracted. It is chosen by the engineer. It is not learned.

## Step 4. Count the updates in an epoch

Lesson 11.3 has 1,000 training examples and a batch size of 50. Each batch gives one update. One epoch is one pass through all the examples. So updates per epoch = 1,000 ÷ 50.

````python
# Step 4: updates per epoch
n_total = 1000
batch_size = 50
updates_per_epoch = n_total // batch_size
print("Updates per epoch:", updates_per_epoch)

epochs = 3
print("Updates after 3 epochs:", epochs * updates_per_epoch)
````

**Expected output**

```
Updates per epoch: 20
Updates after 3 epochs: 60
```

This matches Lesson 11.3: 20 updates per epoch. The second line assumes that each epoch uses the same 20 batches. An epoch counts passes through the data. An update counts changes of the parameters. They are different counts. More training updates do not by themselves mean better performance on unseen data.


With larger batches, one epoch needs fewer updates: 1,000 ÷ 100 = 10. The number of epochs is unchanged. This shows that epochs and updates are different counts.
