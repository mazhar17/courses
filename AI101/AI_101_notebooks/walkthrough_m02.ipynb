## About the data

Lesson 2.3 gives one operating point: Δp = 100 kPa, Q = 0.02 m³/s, efficiency η = 0.8 and a measured input power of 2,800 W. You will check the lesson’s numbers first. Then you will repeat the same calculation for three operating points.

The three points below are synthetic. They were invented for teaching. They are not measurements from a real pump. Point A is the worked example from Lesson 2.3.

```
Point  Δp (kPa)  Q (m³/s)  Measured input power (W)
A           100     0.020                      2800
B            80     0.015                      1650
C           120     0.025                      3950
```


## Step 1. Convert the pressure unit

Lesson 2.1 uses pressure rise Δp in pascals (Pa). The example gives 100 kPa. One kilopascal is 1,000 pascals, so you convert first.

````python
# Step 1: convert kilopascals (kPa) to pascals (Pa)
dp_kpa = 100              # pressure rise in kPa
dp_pa = dp_kpa * 1000     # 1 kPa = 1000 Pa
print(f"Pressure rise: {dp_pa} Pa")
````

**Expected output**

```
Pressure rise: 100000 Pa
```

100 kPa is 100,000 Pa. This matches Lesson 2.1. Without this step, the result would be wrong by a factor of 1,000.

## Step 2. Calculate hydraulic power

The relationship is P_h = Δp × Q. Pa × m³/s gives W, so the unit works out to watts.

````python
# Step 2: hydraulic power, P_h = delta p × Q
q = 0.02                  # volumetric flow rate in m³/s
p_hydraulic = dp_pa * q   # Pa × m³/s = W
print(f"Hydraulic power: {p_hydraulic:.1f} W")
````

**Expected output**

```
Hydraulic power: 2000.0 W
```

This matches the 2,000 W in Lesson 2.1.

## Step 3. Predict the input power

Lesson 2.1 states the input-power relationship P_in = P_h / η. Here η is efficiency. It has no unit. Use η = 0.8, as in the lesson. This is an idealized relationship with one fixed efficiency.

````python
# Step 3: predicted input power, P_in = P_h / efficiency
efficiency = 0.8          # dimensionless
p_input_pred = p_hydraulic / efficiency
print(f"Predicted input power: {p_input_pred:.1f} W")
````

**Expected output**

```
Predicted input power: 2500.0 W
```

This matches the 2,500 W in Lessons 2.1 and 2.3. Hydraulic power (2,000 W) and input power (2,500 W) are different quantities. They are different because of the assumed efficiency.

## Step 4. Calculate the residual

Lesson 2.1 defines the residual as r = y − ŷ, which is the measured value minus the predicted value. Here y is the measured input power of 2,800 W and ŷ is the prediction from Step 3.

````python
# Step 4: residual = measured - predicted
p_measured = 2800         # measured input power in W (from the lesson)
residual = p_measured - p_input_pred
print(f"Residual: {residual:+.1f} W")

if residual > 0:
    print("The measurement is above the prediction.")
else:
    print("The measurement is at or below the prediction.")
````

**Expected output**

```
Residual: +300.0 W
The measurement is above the prediction.
```

r = +300 W, as in Lesson 2.3. A positive residual means the physics prediction is below the measurement. The residual shows a discrepancy. As Lesson 2.1 says, it does not by itself prove the physical cause.

## Step 5. Repeat for three operating points

Now you repeat Steps 1 to 4 for three synthetic points. Each point is a group of four values inside round brackets: a name, Δp in kPa, Q in m³/s and the measured input power in W. The `for` loop takes the points one at a time. The unit conversion is now written inside the loop.

````python
# Step 5: the same calculation for three synthetic operating points
efficiency = 0.8          # dimensionless; you will change this later

points = [
    ("A", 100, 0.020, 2800),
    ("B",  80, 0.015, 1650),
    ("C", 120, 0.025, 3950),
]

print("Point  P_hyd (W)  P_in (W)  Measured (W)  Residual (W)")
for name, dp_kpa, q, p_measured in points:
    p_hydraulic = dp_kpa * 1000 * q
    p_input_pred = p_hydraulic / efficiency
    residual = p_measured - p_input_pred
    print(f"{name:<7}{p_hydraulic:>9.1f}{p_input_pred:>10.1f}{p_measured:>14.1f}{residual:>14.1f}")
````

**Expected output**

```
Point  P_hyd (W)  P_in (W)  Measured (W)  Residual (W)
A         2000.0    2500.0        2800.0         300.0
B         1200.0    1500.0        1650.0         150.0
C         3000.0    3750.0        3950.0         200.0
```

Point A repeats the lesson example. At all three synthetic points the measured power is above the prediction, so every residual is positive. This only suggests that the simple model leaves something out. It does not show why. Three invented points also say nothing about a real pump.

A residual depends on the physics assumptions as well as on the measurement. A hybrid model that learns residuals learns a correction to the baseline you chose.
