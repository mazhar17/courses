## About the data

The heater model from Lesson 7.1 is ΔT = b₀ + b₁P, with b₀ = 1 K and b₁ = 0.6 K/W. The temperature rises and predictions come from the lessons.

The data are illustrative and synthetic. They are not from a real heater.

```
Measured ΔT (K):    5   8  10
Predicted ΔT (K):   4  10   7
Training targets for a baseline (K): 4, 8, 10, 12
```


## Step 1. One prediction and its residual

At P = 10 W, ΔT_predicted = 1 K + 0.6 K/W × 10 W = 7 K. The measured rise is 8 K. The residual is r = y − ŷ, measured minus predicted.

````python
# Step 1: prediction and residual (Lesson 7.1)
b0 = 1.0                  # intercept in K
b1 = 0.6                  # slope in K/W
power = 10.0              # heater power in W

predicted = b0 + b1 * power       # K
measured = 8.0                    # measured temperature rise in K
residual = measured - predicted   # K
print(f"Predicted rise: {predicted:.1f} K")
print(f"Residual: {residual:+.1f} K")
````

**Expected output**

```
Predicted rise: 7.0 K
Residual: +1.0 K
```

The prediction is 7 K and the residual is +1 K, as in Lesson 7.1. The positive sign means the prediction is below the measurement.

## Step 2. Residuals, MAE and RMSE for three observations

Lesson 7.2 gives three measurements, [5, 8, 10] K, and three predictions, [4, 10, 7] K. MAE averages the absolute residuals. RMSE squares the residuals, averages them and takes the square root. Both are in kelvin.

````python
# Step 2: residuals, MAE and RMSE (Lesson 7.2)
measured = [5, 8, 10]     # measured temperature rises in K
predicted = [4, 10, 7]    # predicted temperature rises in K

residuals = []
for i in range(len(measured)):
    residuals.append(measured[i] - predicted[i])
print("Residuals (K):", residuals)

abs_total = 0
sq_total = 0
for r in residuals:
    abs_total += abs(r)
    sq_total += r ** 2
n = len(residuals)
mae = abs_total / n
rmse = (sq_total / n) ** 0.5
print(f"MAE:  {mae:.3f} K")
print(f"RMSE: {rmse:.3f} K")
````

**Expected output**

```
Residuals (K): [1, -2, 3]
MAE:  2.000 K
RMSE: 2.160 K
```

These match Lesson 7.2: residuals [1, −2, 3] K, MAE = 2 K and RMSE ≈ 2.160 K. RMSE is at least as large as MAE, because squaring gives the larger residual (3 K) more influence.

## Step 3. R² compares with predicting the mean

R² = 1 − SSE/SST. SSE is the sum of the squared residuals. SST is the sum of the squared distances of the measurements from their own mean. The squared units cancel, so R² has no unit. A value below zero means the model did worse than predicting the mean.

````python
# Step 3: R-squared (Lesson 7.2)
mean_measured = sum(measured) / n
sse = sq_total                          # sum of squared residuals, in K²
sst = 0
for y in measured:
    sst += (y - mean_measured) ** 2     # in K²

r2 = 1 - sse / sst
print(f"SSE = {sse:.2f} K², SST = {sst:.2f} K²")
print(f"R²  = {r2:.4f}")
````

**Expected output**

```
SSE = 14.00 K², SST = 12.67 K²
R²  = -0.1053
```

R² ≈ −0.1053, as in Lesson 7.2. It is negative because the model’s squared error (14) is larger than the squared error of always predicting the mean (12.67). It is a comparison, not a safety threshold.

## Step 4. Compare with a training-mean baseline

Lesson 7.3 asks you to compare with a training-mean baseline on the same evaluation examples. The baseline predicts the mean of the training targets for every example. Here the training targets are [4, 8, 10, 12] K. The evaluation examples are the same three measurements as in Step 2.

````python
# Step 4: training-mean baseline on the same evaluation examples
train_targets = [4, 8, 10, 12]               # training targets in K
baseline_value = sum(train_targets) / len(train_targets)

baseline_total = 0
for y in measured:
    baseline_total += abs(y - baseline_value)
baseline_mae = baseline_total / len(measured)

print(f"Baseline prediction for every example: {baseline_value:.1f} K")
print(f"Baseline MAE: {baseline_mae:.3f} K")
print(f"Model MAE:    {mae:.3f} K")
if baseline_mae < mae:
    print("The simple baseline has the lower MAE.")
else:
    print("The model has the lower MAE.")
````

**Expected output**

```
Baseline prediction for every example: 8.5 K
Baseline MAE: 1.833 K
Model MAE:    2.000 K
The simple baseline has the lower MAE.
```

The simple baseline beats the model on MAE here. Lesson 7.3 says to compare with a fair baseline and to report the result honestly. With only three evaluation examples, this comparison is also very uncertain.
